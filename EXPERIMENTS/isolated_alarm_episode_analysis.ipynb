```python
import numbers
import pandas as pd
import re

def clean_numeric_text(value):
    if pd.isna(value):
        return pd.NA
    text = str(value).strip().replace('−', '-')
    if text == '':
        return pd.NA
    return text

def parse_mixed_value(value):
    text = clean_numeric_text(value)
    if pd.isna(text):
        return pd.NA
    numeric_candidate = re.sub(r'\s+', '', text).replace(',', '')
    try:
        parsed = pd.to_numeric(numeric_candidate)
    except (TypeError, ValueError):
        return text
    if pd.isna(parsed):
        return text
    return parsed

def is_numeric_value(value):
    return pd.notna(value) and isinstance(value, numbers.Number) and not isinstance(value, bool)

cluster_workbook_path = '/home/h604827/ControlActions/DATA/1071_pvlo_alarms_clustered_with_control_actions.xlsx'
control_action_events_path = '/home/h604827/ControlActions/DATA/trip_filtered_events_dedup.csv'

cluster_sheet_df = pd.read_excel(cluster_workbook_path, sheet_name='alarm_clusters')
control_actions_sheet_df = pd.read_excel(cluster_workbook_path, sheet_name='control_actions')
control_actions_sheet_df['VT_Start'] = pd.to_datetime(control_actions_sheet_df['VT_Start'])
control_actions_sheet_df[['PrevValue', 'Value']] = control_actions_sheet_df[['PrevValue', 'Value']].astype('object')

control_action_events_df = pd.read_csv(control_action_events_path, low_memory=False)
control_action_events_df['VT_Start'] = pd.to_datetime(control_action_events_df['VT_Start'])
change_events_df = control_action_events_df.loc[
    control_action_events_df['ConditionName'] == 'CHANGE',
    ['VT_Start', 'Source', 'Description', 'PrevValue', 'Value']
].copy()
change_events_df['VT_ms'] = change_events_df['VT_Start'].dt.round('ms')

control_actions_sheet_df['VT_ms'] = control_actions_sheet_df['VT_Start'].dt.round('ms')
control_actions_sheet_df = control_actions_sheet_df.merge(
    change_events_df[['VT_ms', 'Source', 'Description', 'PrevValue', 'Value']].rename(
        columns={'PrevValue': 'PrevValue_src', 'Value': 'Value_src'}
    ),
    on=['VT_ms', 'Source', 'Description'],
    how='left'
 )

for col in ['PrevValue', 'Value']:
    restored_values = control_actions_sheet_df[f'{col}_src'].combine_first(control_actions_sheet_df[col])
    control_actions_sheet_df[col] = restored_values.map(parse_mixed_value).astype('object')

control_actions_sheet_df = control_actions_sheet_df.drop(columns=['VT_ms', 'PrevValue_src', 'Value_src'])

numeric_value_mask = control_actions_sheet_df[['PrevValue', 'Value']].apply(
    lambda series: series.map(is_numeric_value)
 )
textual_op_sp_rows = control_actions_sheet_df.loc[
    control_actions_sheet_df['Description'].isin(['OP', 'SP']) & ~numeric_value_mask.all(axis=1)
].copy()

prev_count = control_actions_sheet_df['PrevValue'].notna().sum()
value_count = control_actions_sheet_df['Value'].notna().sum()
if prev_count != value_count:
    raise ValueError(
        f'PrevValue and Value non-null counts differ after restoration: {prev_count} vs {value_count}'
    )

missing_rows = control_actions_sheet_df[['PrevValue', 'Value']].isna().any(axis=1).sum()
if missing_rows:
    raise ValueError(
        f'control_actions contains {missing_rows} rows with missing PrevValue/Value after restoration'
    )
```

```python
# ============================================================
# Extract all alarm events: start (Action=NaN) and end (Action='OK')
# Exclude non-alarm ConditionNames (CHANGE, OPERATOR MESSAGE, MESSAGE, UNCEVT)
# ============================================================
events_df['VT_Start'] = pd.to_datetime(events_df['VT_Start'])

non_alarm_conditions = ['CHANGE', 'OPERATOR MESSAGE', 'MESSAGE', 'UNCEVT']

# Keep only alarm start and end rows, excluding non-alarm condition types
alarm_events = events_df[
    (events_df['Action'].isna() | (events_df['Action'] == 'OK')) &
    (~events_df['ConditionName'].isin(non_alarm_conditions))
][['Source', 'Action', 'VT_Start', 'ConditionName']].copy()

alarm_events = alarm_events.sort_values(['Source', 'VT_Start']).reset_index(drop=True)
print(f"Total alarm start/end events: {len(alarm_events)}")
print(f"  Starts (Action=NaN): {alarm_events['Action'].isna().sum()}")
print(f"  Ends (Action=OK): {(alarm_events['Action'] == 'OK').sum()}")
print(f"  Unique sources: {alarm_events['Source'].nunique()}")
print(f"\nConditionName breakdown (starts only):")
print(alarm_events[alarm_events['Action'].isna()]['ConditionName'].value_counts())
```

```python
# ============================================================
# Pair alarm starts with their next OK (per Source + ConditionName)
# State-machine approach: track alarm on/off state
# ============================================================
# For each (Source, ConditionName):
#   - First NaN when alarm is OFF → alarm starts
#   - First OK when alarm is ON → alarm ends
#   - Additional NaNs while alarm is ON → re-triggers, skip
#   - OKs while alarm is OFF → stale, skip

paired_episodes = []

for (source, cond), grp in alarm_events.groupby(['Source', 'ConditionName']):
    grp = grp.sort_values('VT_Start')
    alarm_on = False
    start_time = None
    
    for _, row in grp.iterrows():
        if not alarm_on and pd.isna(row['Action']):
            # Alarm starts
            alarm_on = True
            start_time = row['VT_Start']
        elif alarm_on and row['Action'] == 'OK':
            # Alarm ends
            paired_episodes.append({
                'Source': source,
                'ConditionName': cond,
                'alarm_start': start_time,
                'alarm_end': row['VT_Start']
            })
            alarm_on = False
            start_time = None

all_alarm_episodes = pd.DataFrame(paired_episodes)
all_alarm_episodes['alarm_start'] = pd.to_datetime(all_alarm_episodes['alarm_start'])
all_alarm_episodes['alarm_end'] = pd.to_datetime(all_alarm_episodes['alarm_end'])

print(f"Total alarm episodes (all tags): {len(all_alarm_episodes)}")
print(f"Unique tags with alarm episodes: {all_alarm_episodes['Source'].nunique()}")
print(f"\nTop 10 tags by number of alarm episodes:")
print(all_alarm_episodes['Source'].value_counts().head(10))

# Quick sanity check on durations
durations = (all_alarm_episodes['alarm_end'] - all_alarm_episodes['alarm_start']).dt.total_seconds() / 3600
print(f"\nAlarm duration stats (hours) across all tags:")
print(durations.describe())
```

```python
# ============================================================
# For each 1071 alarm cluster, find overlapping alarms from other tags
# ============================================================
# An alarm from another tag overlaps with a 1071 cluster if:
#   other_alarm_start < cluster_end AND other_alarm_end > cluster_start

# Exclude 03LIC_1071 from the other alarms
other_alarms = all_alarm_episodes[all_alarm_episodes['Source'] != '03LIC_1071'].copy()
other_starts = other_alarms['alarm_start'].values
other_ends = other_alarms['alarm_end'].values
other_sources = other_alarms['Source'].values

results = []

for idx, row in alarm_episodes_1071.iterrows():
    ep_start = row['cluster_start_time']
    ep_end = row['cluster_end_time']
    
    mask = (other_starts < ep_end) & (other_ends > ep_start)
    overlapping = other_alarms[mask]
    overlapping_sources = sorted(overlapping['Source'].unique().tolist())
    
    results.append({
        'episode_idx': idx,
        'cluster_id': row['cluster_id'],
        'alarm_start': ep_start,
        'alarm_end': ep_end,
        'cluster_total_alarms': row['cluster_total_alarms'],
        'cluster_type': row['cluster_type'],
        'num_overlapping_alarms': len(overlapping_sources),
        'overlapping_tags': overlapping_sources
    })

overlap_df = pd.DataFrame(results)
print(f"Total 1071 alarm clusters: {len(overlap_df)}")
print(f"Clusters with NO overlapping alarms: {(overlap_df['num_overlapping_alarms'] == 0).sum()}")
print(f"Clusters WITH overlapping alarms: {(overlap_df['num_overlapping_alarms'] > 0).sum()}")
print(f"\nDistribution of overlapping alarm count:")
print(overlap_df['num_overlapping_alarms'].describe())
```

```python
# ============================================================
# Which tags most frequently overlap with 1071 alarm clusters?
# ============================================================
from collections import Counter

tag_counter = Counter()
for tags in overlap_df['overlapping_tags']:
    tag_counter.update(tags)

tag_freq_df = pd.DataFrame(tag_counter.most_common(), columns=['Tag', 'Episodes_Overlapping'])
tag_freq_df['Pct_of_clusters'] = (tag_freq_df['Episodes_Overlapping'] / len(overlap_df) * 100).round(1)

print(f"Total unique tags that overlap with 1071 alarm clusters: {len(tag_freq_df)}")
print(f"\nTags that overlap with all {len(overlap_df)} alarm clusters:")
always_on = tag_freq_df[tag_freq_df['Episodes_Overlapping'] == len(overlap_df)]
print(f"  {len(always_on)} tags" if len(always_on) > 0 else "  None")
if len(always_on) > 0:
    print(always_on.to_string(index=False))

print(f"\nTop 20 most frequently overlapping tags:")
tag_freq_df.head(20)
```

```python
# ============================================================
# Per-cluster detail: filter overlapping and operated tags to only those in paths
# ============================================================
# Create set of all tags from paths
path_tags = set()
for tag_list in paths.values():
    path_tags.update(tag_list)

display_df = overlap_df[[
    'episode_idx', 'cluster_id', 'alarm_start', 'alarm_end',
    'cluster_total_alarms', 'cluster_type', 'num_overlapping_alarms', 'overlapping_tags'
]].copy()

# Filter overlapping tags to only those present in paths
display_df['overlapping_tags_filtered'] = display_df['overlapping_tags'].apply(
    lambda tags: [t for t in tags if t in path_tags]
)
display_df['num_overlapping_filtered'] = display_df['overlapping_tags_filtered'].apply(len)
display_df['filtered_tags_str'] = display_df['overlapping_tags_filtered'].apply(lambda x: ', '.join(x))

# Get operated tags from the control_actions sheet for each cluster
operated_tags_by_cluster = (
    control_actions_sheet_df.groupby('cluster_id')['Source']
    .agg(lambda values: sorted({value for value in values if pd.notna(value)}))
    .to_dict()
 )

display_df['operated_tags'] = display_df['cluster_id'].map(
    lambda cluster_id: operated_tags_by_cluster.get(cluster_id, [])
)
display_df['num_operated_tags'] = display_df['operated_tags'].apply(len)
display_df['operated_tags_filtered'] = display_df['operated_tags'].apply(
    lambda tags: [t for t in tags if t in path_tags]
)
display_df['num_operated_tags_filtered'] = display_df['operated_tags_filtered'].apply(len)
display_df['operated_tags_filtered_str'] = display_df['operated_tags_filtered'].apply(
    lambda x: ', '.join(x)
)

display_df['duration_min'] = (display_df['alarm_end'] - display_df['alarm_start']).dt.total_seconds() / 60

print(f"Path tags: {len(path_tags)} unique tags")
print(f"Before filtering: {overlap_df['num_overlapping_alarms'].sum()} total overlaps")
print(f"After filtering to path tags: {display_df['num_overlapping_filtered'].sum()} total overlaps")
print(f"Clusters with operated tags: {(display_df['num_operated_tags'] > 0).sum()} / {len(display_df)}")
print(f"Clusters with path-filtered operated tags: {(display_df['num_operated_tags_filtered'] > 0).sum()} / {len(display_df)}")

print("\nAlarm clusters with fewest path-filtered overlapping alarms:")
display_df.sort_values(['num_overlapping_filtered', 'cluster_id']).head(20)[
    ['cluster_id', 'alarm_start', 'alarm_end', 'duration_min',
     'cluster_total_alarms', 'num_overlapping_alarms', 'num_overlapping_filtered',
     'num_operated_tags', 'num_operated_tags_filtered']
]
```

```python
# ============================================================
# Append overlap results to the cluster workbook and restore control_actions values
# Convert only values that are actually numeric-looking; preserve text even on OP/SP rows
# ============================================================
export_df = display_df[[
    'episode_idx', 'cluster_id', 'alarm_start', 'alarm_end', 'duration_min',
    'cluster_total_alarms', 'cluster_type',
    'num_overlapping_alarms', 'num_overlapping_filtered',
    'num_operated_tags', 'num_operated_tags_filtered',
    'filtered_tags_str', 'operated_tags_filtered_str',
]].copy()

prev_count = control_actions_sheet_df['PrevValue'].notna().sum()
value_count = control_actions_sheet_df['Value'].notna().sum()
if prev_count != value_count:
    raise ValueError(
        f'PrevValue and Value non-null counts differ before save: {prev_count} vs {value_count}'
    )

missing_rows = control_actions_sheet_df[['PrevValue', 'Value']].isna().any(axis=1).sum()
if missing_rows:
    raise ValueError(
        f'control_actions contains {missing_rows} rows with missing PrevValue/Value before save'
    )

numeric_value_mask = control_actions_sheet_df[['PrevValue', 'Value']].apply(
    lambda series: series.map(is_numeric_value)
 )
textual_op_sp_rows = control_actions_sheet_df.loc[
    control_actions_sheet_df['Description'].isin(['OP', 'SP']) & ~numeric_value_mask.all(axis=1)
].copy()

numeric_prev_count = int(numeric_value_mask['PrevValue'].sum())
numeric_value_count = int(numeric_value_mask['Value'].sum())
textual_op_sp_count = len(textual_op_sp_rows)

with pd.ExcelWriter(
    cluster_workbook_path, engine='openpyxl', mode='a', if_sheet_exists='replace'
 ) as writer:
    control_actions_sheet_df.to_excel(writer, sheet_name='control_actions', index=False)
    export_df.to_excel(writer, sheet_name='overlapping_alarms', index=False)

print(f"Updated workbook: {cluster_workbook_path}")
print(f"  - control_actions rows: {len(control_actions_sheet_df)}")
print(f"  - PrevValue non-null count: {prev_count}")
print(f"  - Value non-null count: {value_count}")
print(f"  - Numeric-looking PrevValue cells: {numeric_prev_count}")
print(f"  - Numeric-looking Value cells: {numeric_value_count}")
print(f"  - OP/SP rows with preserved text values: {textual_op_sp_count}")
print(f"  - overlapping_alarms: {len(export_df)} rows")
```
