```python
import pandas as pd

events_df = pd.read_csv('/home/h604827/ControlActions/DATA/trip_filtered_events.csv')
ssd_df = pd.read_excel('/home/h604827/ControlActions/DATA/SSD_1071_SSD_output_1071_7Jan2026.xlsx')
events_df.head()
```

```python
print("SSD columns:", ssd_df.columns.tolist())
print("\nSSD shape:", ssd_df.shape)
print("\nSSD head:")
ssd_df.head(3)
```

```python
# Get unique 1071 alarm episodes from SSD data
alarm_episodes_1071 = ssd_df[['AlarmStart_rounded_minutes', 'AlarmEnd_rounded_minutes']].drop_duplicates().reset_index(drop=True)
alarm_episodes_1071['AlarmStart_rounded_minutes'] = pd.to_datetime(alarm_episodes_1071['AlarmStart_rounded_minutes'])
alarm_episodes_1071['AlarmEnd_rounded_minutes'] = pd.to_datetime(alarm_episodes_1071['AlarmEnd_rounded_minutes'])
alarm_episodes_1071 = alarm_episodes_1071.sort_values('AlarmStart_rounded_minutes').reset_index(drop=True)
alarm_episodes_1071.index.name = 'episode_idx'
print(f"Total 1071 alarm episodes: {len(alarm_episodes_1071)}")
alarm_episodes_1071.head()
```

```python
# Explore events data: alarm-related ConditionNames and Action values
print("Unique Action values:", events_df['Action'].unique())
print("\nUnique ConditionName values:", events_df['ConditionName'].unique())
print("\nSample alarm events (Action is null):")
events_df[events_df['Action'].isna()][['Source', 'Action', 'ConditionName', 'VT_Start', 'Category']].head(5)
```

```python
# ============================================================
# Extract all alarm events: start (Action=NaN) and end (Action='OK')
# Exclude non-alarm ConditionNames (CHANGE, OPERATOR MESSAGE, MESSAGE, UNCEVT)
# ============================================================
events_df['VT_Start'] = pd.to_datetime(events_df['VT_Start'])

non_alarm_conditions = ['CHANGE', 'OPERATOR MESSAGE', 'MESSAGE', 'UNCEVT']

# Keep only alarm start and end rows, excluding non-alarm condition types
alarm_events = events_df[
    (events_df['Action'].isna() | (events_df['Action'] == 'OK')) &
    (~events_df['ConditionName'].isin(non_alarm_conditions))
][['Source', 'Action', 'VT_Start', 'ConditionName']].copy()

alarm_events = alarm_events.sort_values(['Source', 'VT_Start']).reset_index(drop=True)
print(f"Total alarm start/end events: {len(alarm_events)}")
print(f"  Starts (Action=NaN): {alarm_events['Action'].isna().sum()}")
print(f"  Ends (Action=OK): {(alarm_events['Action'] == 'OK').sum()}")
print(f"  Unique sources: {alarm_events['Source'].nunique()}")
print(f"\nConditionName breakdown (starts only):")
print(alarm_events[alarm_events['Action'].isna()]['ConditionName'].value_counts())
```

```python
# ============================================================
# Pair alarm starts with their next OK (per Source + ConditionName)
# State-machine approach: track alarm on/off state
# ============================================================
# For each (Source, ConditionName):
#   - First NaN when alarm is OFF → alarm starts
#   - First OK when alarm is ON → alarm ends
#   - Additional NaNs while alarm is ON → re-triggers, skip
#   - OKs while alarm is OFF → stale, skip

paired_episodes = []

for (source, cond), grp in alarm_events.groupby(['Source', 'ConditionName']):
    grp = grp.sort_values('VT_Start')
    alarm_on = False
    start_time = None
    
    for _, row in grp.iterrows():
        if not alarm_on and pd.isna(row['Action']):
            # Alarm starts
            alarm_on = True
            start_time = row['VT_Start']
        elif alarm_on and row['Action'] == 'OK':
            # Alarm ends
            paired_episodes.append({
                'Source': source,
                'ConditionName': cond,
                'alarm_start': start_time,
                'alarm_end': row['VT_Start']
            })
            alarm_on = False
            start_time = None

all_alarm_episodes = pd.DataFrame(paired_episodes)
all_alarm_episodes['alarm_start'] = pd.to_datetime(all_alarm_episodes['alarm_start'])
all_alarm_episodes['alarm_end'] = pd.to_datetime(all_alarm_episodes['alarm_end'])

print(f"Total alarm episodes (all tags): {len(all_alarm_episodes)}")
print(f"Unique tags with alarm episodes: {all_alarm_episodes['Source'].nunique()}")
print(f"\nTop 10 tags by number of alarm episodes:")
print(all_alarm_episodes['Source'].value_counts().head(10))

# Quick sanity check on durations
durations = (all_alarm_episodes['alarm_end'] - all_alarm_episodes['alarm_start']).dt.total_seconds() / 3600
print(f"\nAlarm duration stats (hours) across all tags:")
print(durations.describe())
```

```python
# ============================================================
# For each 1071 alarm episode, find overlapping alarms from other tags
# ============================================================
# An alarm from another tag overlaps with a 1071 episode if:
#   other_alarm_start < episode_1071_end AND other_alarm_end > episode_1071_start

# Exclude 03LIC_1071 from the other alarms
other_alarms = all_alarm_episodes[all_alarm_episodes['Source'] != '03LIC_1071'].copy()
other_starts = other_alarms['alarm_start'].values
other_ends = other_alarms['alarm_end'].values
other_sources = other_alarms['Source'].values

results = []

for idx, row in alarm_episodes_1071.iterrows():
    ep_start = row['AlarmStart_rounded_minutes']
    ep_end = row['AlarmEnd_rounded_minutes']
    
    # Find overlapping: other starts before ep ends AND other ends after ep starts
    mask = (other_starts < ep_end) & (other_ends > ep_start)
    overlapping = other_alarms[mask]
    
    overlapping_sources = sorted(overlapping['Source'].unique().tolist())
    
    results.append({
        'episode_idx': idx,
        'alarm_start': ep_start,
        'alarm_end': ep_end,
        'num_overlapping_alarms': len(overlapping_sources),
        'overlapping_tags': overlapping_sources
    })

overlap_df = pd.DataFrame(results)
print(f"Total 1071 episodes: {len(overlap_df)}")
print(f"Episodes with NO overlapping alarms (isolated): {(overlap_df['num_overlapping_alarms'] == 0).sum()}")
print(f"Episodes WITH overlapping alarms: {(overlap_df['num_overlapping_alarms'] > 0).sum()}")
print(f"\nDistribution of overlapping alarm count:")
print(overlap_df['num_overlapping_alarms'].describe())
```

```python
# ============================================================
# Which tags most frequently overlap with 1071 alarms?
# ============================================================
from collections import Counter

tag_counter = Counter()
for tags in overlap_df['overlapping_tags']:
    tag_counter.update(tags)

tag_freq_df = pd.DataFrame(tag_counter.most_common(), columns=['Tag', 'Episodes_Overlapping'])
tag_freq_df['Pct_of_609_episodes'] = (tag_freq_df['Episodes_Overlapping'] / len(overlap_df) * 100).round(1)

print(f"Total unique tags that overlap with 1071 alarms: {len(tag_freq_df)}")
print(f"\nTags that overlap with ALL 609 episodes:")
always_on = tag_freq_df[tag_freq_df['Episodes_Overlapping'] == 609]
print(f"  {len(always_on)} tags" if len(always_on) > 0 else "  None")
if len(always_on) > 0:
    print(always_on.to_string(index=False))

print(f"\nTop 20 most frequently overlapping tags:")
tag_freq_df.head(20)
```

```python
# ============================================================
# Identify "chronic" alarms (always on) vs meaningful co-occurring alarms
# ============================================================
# Tags that are in alarm for >95% of 1071 episodes are likely chronic/nuisance alarms
CHRONIC_THRESHOLD = 0.95
chronic_tags = set(tag_freq_df[tag_freq_df['Pct_of_609_episodes'] > CHRONIC_THRESHOLD * 100]['Tag'])
print(f"Chronic alarm tags (>{CHRONIC_THRESHOLD*100}% overlap): {len(chronic_tags)}")
for t in sorted(chronic_tags):
    pct = tag_freq_df[tag_freq_df['Tag'] == t]['Pct_of_609_episodes'].values[0]
    print(f"  {t}: {pct}%")

# Recompute overlap excluding chronic tags
overlap_df['overlapping_tags_filtered'] = overlap_df['overlapping_tags'].apply(
    lambda tags: [t for t in tags if t not in chronic_tags]
)
overlap_df['num_overlapping_filtered'] = overlap_df['overlapping_tags_filtered'].apply(len)

print(f"\n--- After removing chronic alarms ---")
print(f"Isolated episodes (no other non-chronic alarms): {(overlap_df['num_overlapping_filtered'] == 0).sum()}")
print(f"Episodes with non-chronic overlapping alarms: {(overlap_df['num_overlapping_filtered'] > 0).sum()}")
print(f"\nDistribution of non-chronic overlapping alarm count:")
print(overlap_df['num_overlapping_filtered'].describe())
```

```python
# ============================================================
# Per-episode detail: show overlapping tags for each 1071 episode
# ============================================================
# Show episodes sorted by fewest overlapping alarms (most isolated first)
display_df = overlap_df[['episode_idx', 'alarm_start', 'alarm_end', 
                          'num_overlapping_alarms', 'num_overlapping_filtered',
                          'overlapping_tags_filtered']].copy()
display_df = display_df.rename(columns={
    'num_overlapping_alarms': 'total_overlapping',
    'num_overlapping_filtered': 'non_chronic_overlapping',
    'overlapping_tags_filtered': 'non_chronic_tags'
})
display_df['duration_min'] = (display_df['alarm_end'] - display_df['alarm_start']).dt.total_seconds() / 60
display_df['non_chronic_tags_str'] = display_df['non_chronic_tags'].apply(lambda x: ', '.join(x))

# Show the most "isolated" episodes (fewest non-chronic overlapping alarms)
print("Episodes with fewest non-chronic overlapping alarms:")
display_df.sort_values('non_chronic_overlapping').head(20)[
    ['episode_idx', 'alarm_start', 'alarm_end', 'duration_min', 
     'total_overlapping', 'non_chronic_overlapping', 'non_chronic_tags_str']
]
```

```python
# ============================================================
# Save full results to Excel
# ============================================================
# Sheet 1: Per-episode overview with overlapping tags
export_df = display_df[['episode_idx', 'alarm_start', 'alarm_end', 'duration_min',
                          'total_overlapping', 'non_chronic_overlapping', 'non_chronic_tags_str']].copy()

# Sheet 2: Tag frequency table
with pd.ExcelWriter('/home/h604827/ControlActions/RESULTS/1071_overlapping_alarms.xlsx') as writer:
    export_df.to_excel(writer, sheet_name='Per_Episode', index=False)
    tag_freq_df.to_excel(writer, sheet_name='Tag_Frequency', index=False)
    
    # Sheet 3: Chronic tags
    chronic_df = tag_freq_df[tag_freq_df['Pct_of_609_episodes'] > CHRONIC_THRESHOLD * 100]
    chronic_df.to_excel(writer, sheet_name='Chronic_Tags', index=False)

print("Saved to RESULTS/1071_overlapping_alarms.xlsx")
print(f"  - Per_Episode: {len(export_df)} rows")
print(f"  - Tag_Frequency: {len(tag_freq_df)} tags")
print(f"  - Chronic_Tags: {len(chronic_df)} tags")
```

```python
# ============================================================
# Inspect overlapping alarms for a specific tag
# ============================================================
TAG_TO_INSPECT = '03LIC_1016'  # <-- Change this to any tag from the 263

# Find all alarm episodes for this tag that overlap with any 1071 episode
tag_episodes = all_alarm_episodes[all_alarm_episodes['Source'] == TAG_TO_INSPECT].copy()

overlapping_pairs = []
for _, ep1071 in alarm_episodes_1071.iterrows():
    ep_start = ep1071['AlarmStart_rounded_minutes']
    ep_end = ep1071['AlarmEnd_rounded_minutes']
    
    mask = (tag_episodes['alarm_start'] < ep_end) & (tag_episodes['alarm_end'] > ep_start)
    for _, tag_ep in tag_episodes[mask].iterrows():
        overlapping_pairs.append({
            '1071_episode_idx': ep1071.name,
            '1071_alarm_start': ep_start,
            '1071_alarm_end': ep_end,
            f'{TAG_TO_INSPECT}_alarm_start': tag_ep['alarm_start'],
            f'{TAG_TO_INSPECT}_alarm_end': tag_ep['alarm_end'],
        })

pairs_df = pd.DataFrame(overlapping_pairs).drop_duplicates()
print(f"Tag: {TAG_TO_INSPECT}")
print(f"Total alarm episodes for this tag: {len(tag_episodes)}")
print(f"Overlapping pairs with 1071 episodes: {len(pairs_df)}")
print(f"\nShowing first 20:")
pairs_df.head(20)
```
