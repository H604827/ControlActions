# Historical Operator-Action Constraints — 03LIC_1071 PVLO Alarms

Derive **historical constraints** on operator control actions, so any automated
Control-Actions recommendation can be bounded to what operators have actually done.

## Two constraint families

1. **OP value range (absolute bounds)** — from the PV/OP parquet.
   For each action tag's `.OP` column, the envelope within which the OP value has
   historically been kept (hard `min`/`max` plus a robust **p1–p99** band).

2. **Step-change range (per action)** — from the alarm control-actions workbook.
   For each action tag, the range of single-move step changes
   `Step = Value − PrevValue`, split by `Description` (**OP** vs **SP**).

All tag-level tables are **sorted by how often the tag was operated** (most-operated first).

## Data sources
- **Actions**: `RESULTS/03LIC_1071_PVLO_episodes_12JUN2026_1219/03LIC_1071_pvlo_alarms_clustered_with_control_actions.xlsx`
  (`control_actions` sheet, ~42k rows, 195 tags). Rows span `[alarm_start − 4h,
  alarm_end + 1h]` for every 1071 PVLO alarm. `Description` ∈ {OP, SP, MODE, SO, FL(..), …}; `Source` = tag.
- **PV/OP history**: `DATA/PV-OP_data/03LIC_1071_JAN_2026.parquet`
  (15 `.OP` columns, 1-min, 2022–2025).

## Notes / caveats
- **De-duplication**: the workbook is cluster-scoped, so overlapping alarm windows
  repeat the same physical action under several `cluster_id`s (up to 4×). Actions are
  collapsed to unique `(Source, Description, VT_Start, PrevValue, Value)` before any stats,
  otherwise counts and percentiles would be inflated.
- **OP steps are in %** → comparable across tags. **SP steps are in each tag's
  engineering units** → *not* comparable across tags (kept in a separate table).
- Non-numeric OP moves (OPEN/CLOSE, discrete valves) and MODE / SO / FL(..) events
  are excluded from step math.
- Raw OP `min`/`max` can include shutdown / sensor artifacts (e.g. negative %); the
  robust **p1–p99** band is the meaningful operating envelope.
- The `filtered` column flags actions in Stage-4 "passing" clusters (a curated
  subset); **all** clusters are used here by default (the passing copy is kept on dedup).


```python
from pathlib import Path

import numpy as np
import pandas as pd
import plotly.express as px
import pyarrow.parquet as pq_mod

pd.set_option("display.width", 220)
pd.set_option("display.max_columns", 60)


def find_repo() -> Path:
    """Locate the ControlActions repo root from any working directory."""
    here = Path.cwd()
    for cand in [here, *here.parents]:
        if (cand / "DATA" / "PV-OP_data").exists() and (cand / "RESULTS").exists():
            return cand
    return here


REPO = find_repo()
ACTIONS_XLSX = REPO / ("RESULTS/03LIC_1071_PVLO_episodes_12JUN2026_1219/"
                       "03LIC_1071_pvlo_alarms_clustered_with_control_actions.xlsx")
PV_PARQUET = REPO / "DATA/PV-OP_data/03LIC_1071_JAN_2026.parquet"

OUT_DIR = REPO / "RESULTS/historical_action_constraints"
OUT_DIR.mkdir(parents=True, exist_ok=True)
OUT_XLSX = OUT_DIR / "historical_action_constraints_1071.xlsx"

OP_VALID_LO, OP_VALID_HI = 0.0, 105.0  # physically plausible OP band (%)

print("Repo root       :", REPO)
print("Actions workbook:", ACTIONS_XLSX.name, "| exists:", ACTIONS_XLSX.exists())
print("PV/OP parquet   :", PV_PARQUET.name, "| exists:", PV_PARQUET.exists())
print("Output workbook :", OUT_XLSX)
```

```python
def range_stats(values, prefix: str = "") -> dict:
    """Count / min / robust percentiles / max / mean / std of a numeric series."""
    s = pd.to_numeric(pd.Series(values), errors="coerce").dropna()
    if s.empty:
        return {f"{prefix}n": 0}
    q = s.quantile([0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99])
    return {
        f"{prefix}n": int(s.size),
        f"{prefix}min": round(float(s.min()), 4),
        f"{prefix}p1": round(float(q.loc[0.01]), 4),
        f"{prefix}p5": round(float(q.loc[0.05]), 4),
        f"{prefix}p25": round(float(q.loc[0.25]), 4),
        f"{prefix}p50": round(float(q.loc[0.50]), 4),
        f"{prefix}p75": round(float(q.loc[0.75]), 4),
        f"{prefix}p95": round(float(q.loc[0.95]), 4),
        f"{prefix}p99": round(float(q.loc[0.99]), 4),
        f"{prefix}max": round(float(s.max()), 4),
        f"{prefix}mean": round(float(s.mean()), 4),
        f"{prefix}std": round(float(s.std()), 4),
    }
```

## 1. Load control actions

Load the cluster-scoped workbook and **collapse cluster-overlap duplicates** to unique
physical actions. `Step = Value − PrevValue`, computed only where both are numeric.
Non-numeric moves (MODE, SO, FL(..), OPEN/CLOSE valves) yield `NaN` steps and are
dropped from the step analysis below.


```python
ca_raw = pd.read_excel(ACTIONS_XLSX, sheet_name="control_actions")
ca_raw["VT_Start"] = pd.to_datetime(ca_raw["VT_Start"])

# The workbook is cluster-scoped: overlapping alarm windows (±4h/±1h) repeat the
# same physical action under several cluster_ids. Collapse to unique actions so
# counts / percentiles are not inflated. When a duplicated action appears in both a
# "passing" (filtered=True) and a non-passing cluster, keep the passing copy.
ACTION_KEY = ["Source", "Description", "VT_Start", "PrevValue", "Value"]
if "filtered" in ca_raw.columns:
    ca_raw = ca_raw.sort_values("filtered", ascending=False)
ca = ca_raw.drop_duplicates(subset=ACTION_KEY, keep="first").reset_index(drop=True)

ca["PrevValue_num"] = pd.to_numeric(ca["PrevValue"], errors="coerce")
ca["Value_num"] = pd.to_numeric(ca["Value"], errors="coerce")
ca["step_change"] = ca["Value_num"] - ca["PrevValue_num"]

print(f"Rows in workbook : {len(ca_raw):,}")
print(f"Unique actions   : {len(ca):,} "
      f"(removed {len(ca_raw) - len(ca):,} cluster-overlap duplicates)")
print(f"Unique tags      : {ca['Source'].nunique()} | clusters: {ca['cluster_id'].nunique()}")
print("\nTop action types (Description):\n", ca["Description"].value_counts().head(8).to_string())
print("\nBy action_timing:\n", ca["action_timing"].value_counts().to_string())

numeric = ca.dropna(subset=["step_change"])
print(f"\nNumeric step actions: {len(numeric):,} "
      f"| OP: {int((numeric['Description'] == 'OP').sum()):,} "
      f"| SP: {int((numeric['Description'] == 'SP').sum()):,}")
ca.head()
```

## 2. OP value range (absolute bounds) — from parquet

The envelope within which each `.OP` value has historically stayed. `n_op_actions` is
how many OP moves that tag received in the workbook (table sorted by it, descending);
`n_op_samples` is the number of 1-min parquet readings behind the stats. `min`/`max`
are hard bounds (may include shutdown/artifact extremes — see `n_below_0`); **p1–p99**
is the robust operating band. `is_op_action_tag` flags tags actually moved via OP.


```python
op_cols = sorted(c for c in pq_mod.ParquetFile(PV_PARQUET).schema_arrow.names
                 if c.endswith(".OP"))
op_tags = [c[:-3] for c in op_cols]
print(f"{len(op_cols)} OP columns in parquet")

op_hist = pd.read_parquet(PV_PARQUET, columns=op_cols)

# how many OP moves each tag received in the workbook (deduped)
op_action_counts = ca[ca["Description"] == "OP"].groupby("Source").size()
op_action_tags = set(op_action_counts.index)

rows = []
for tag, col in zip(op_tags, op_cols):
    s = op_hist[col]
    d = {"tag": tag, "op_col": col,
         "is_op_action_tag": tag in op_action_tags,
         "n_op_actions": int(op_action_counts.get(tag, 0))}
    d.update(range_stats(s))
    d["n_below_0"] = int((s < OP_VALID_LO).sum())
    d["n_above_105"] = int((s > OP_VALID_HI).sum())
    rows.append(d)

# 'n' from range_stats is the number of parquet samples -> rename for clarity, and
# sort tags by how often operators actually moved them (most-operated first).
op_value_range = (pd.DataFrame(rows)
                  .rename(columns={"n": "n_op_samples"})
                  .sort_values("n_op_actions", ascending=False)
                  .reset_index(drop=True))
front = ["tag", "op_col", "is_op_action_tag", "n_op_actions", "n_op_samples"]
op_value_range = op_value_range[front + [c for c in op_value_range.columns if c not in front]]
op_value_range
```

## 3. Step-change range per action tag — from workbook

Both tables are **sorted by `n`** (number of step moves for that tag), most-operated first.
`OP` steps are in **%** (comparable across tags). `SP` steps are in each tag's
**engineering units** (NOT comparable — e.g. a flow SP moves in thousands while a
level SP moves in %). They are reported in separate tables. `max_increase` /
`max_decrease` are the largest single up / down moves ever applied. Treat small-`n`
tags as weak/indicative constraints.


```python
def step_range_by_tag(df: pd.DataFrame, description: str) -> pd.DataFrame:
    """Per-tag step-change range for a given Description (OP or SP)."""
    sub = df[df["Description"] == description].dropna(subset=["step_change"])
    out = []
    for tag, g in sub.groupby("Source"):
        step = g["step_change"]
        inc, dec = step[step > 0], step[step < 0]
        d = {"tag": tag}
        d.update(range_stats(step))
        d["n_increase"] = int(inc.size)
        d["n_decrease"] = int(dec.size)
        d["n_zero"] = int((step == 0).sum())
        d["max_increase"] = round(float(inc.max()), 4) if inc.size else np.nan
        d["max_decrease"] = round(float(dec.min()), 4) if dec.size else np.nan
        d["median_abs_step"] = round(float(step.abs().median()), 4)
        out.append(d)
    front = ["tag", "n", "min", "max", "p5", "p50", "p95",
             "n_increase", "n_decrease", "max_increase", "max_decrease", "median_abs_step"]
    res = pd.DataFrame(out)
    ordered = front + [c for c in res.columns if c not in front]
    return res[ordered].sort_values("n", ascending=False).reset_index(drop=True)


op_step_range = step_range_by_tag(ca, "OP")
sp_step_range = step_range_by_tag(ca, "SP")

print("OP step-change range by tag (%, comparable):")
display(op_step_range)
print("\nSP step-change range by tag (engineering units — NOT comparable across tags):")
display(sp_step_range)
```

## 4. Combined OP constraints per action tag

Absolute OP bounds (from parquet) joined with OP step-change bounds (from the
workbook). Tags that took OP moves but have no `.OP` column in the parquet
(e.g. `03HIC_1092A`) get `NaN` value-bounds — step bounds still apply.

```python
op_bounds = op_value_range.rename(columns={
    "min": "op_hard_min", "max": "op_hard_max",
    "p1": "op_band_lo", "p99": "op_band_hi", "p50": "op_median",
})[["tag", "op_hard_min", "op_band_lo", "op_median", "op_band_hi", "op_hard_max"]]

op_moves = op_step_range.rename(columns={
    "n": "n_op_steps", "min": "op_step_min", "max": "op_step_max",
    "p5": "op_step_p5", "p95": "op_step_p95",
    "max_increase": "op_max_up", "max_decrease": "op_max_down",
})[["tag", "n_op_steps", "op_step_min", "op_step_p5", "op_step_p95",
    "op_step_max", "op_max_up", "op_max_down"]]

# most-operated tags first
combined_op_constraints = (op_moves.merge(op_bounds, on="tag", how="left")
                           .sort_values("n_op_steps", ascending=False)
                           .reset_index(drop=True))
combined_op_constraints
```

## 5. Visualizations

```python
# --- OP operating bands (sampled parquet OP, plausible band only) ---
action_op_cols = [f"{t}.OP" for t in sorted(op_action_tags) if f"{t}.OP" in op_cols]
sample = op_hist[action_op_cols].dropna(how="all")
sample = sample.sample(min(40000, len(sample)), random_state=0)
long = sample.melt(var_name="tag", value_name="OP").dropna()
long = long[(long["OP"] >= OP_VALID_LO) & (long["OP"] <= OP_VALID_HI)]

fig = px.box(long, x="tag", y="OP",
             title="Historical OP value range by action tag (plausible band, sampled)")
fig.update_layout(height=430, xaxis_title="", yaxis_title="OP value (%)")
fig.show()
```

```python
# --- OP step-change range (%), top-15 most-operated tags, ordered by move count ---
top_op = op_step_range["tag"].head(15).tolist()
op_long = ca[(ca["Description"] == "OP") & ca["Source"].isin(top_op)]
op_long = op_long.dropna(subset=["step_change"])[["Source", "step_change"]]
fig = px.box(op_long, x="Source", y="step_change", points="outliers",
             category_orders={"Source": top_op},
             title="OP step-change range — top 15 most-operated tags (%)")
fig.add_hline(y=0, line_dash="dot", line_color="gray")
fig.update_layout(height=450, xaxis_title="", yaxis_title="Step = Value − PrevValue (%)")
fig.show()
```

```python
# --- SP step-change range (engineering units), top-12 tags, independent y per tag ---
top_sp = sp_step_range["tag"].head(12).tolist()
sp_long = ca[(ca["Description"] == "SP") & ca["Source"].isin(top_sp)]
sp_long = sp_long.dropna(subset=["step_change"])[["Source", "step_change"]]
fig = px.box(sp_long, y="step_change", facet_col="Source", facet_col_wrap=4, points="outliers",
             category_orders={"Source": top_sp},
             title="SP step-change range — top 12 tags (engineering units — independent y per tag)")
fig.update_yaxes(matches=None)
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
fig.update_layout(height=560)
fig.show()
```

## 6. Save constraints workbook

```python
with pd.ExcelWriter(OUT_XLSX, engine="openpyxl") as xw:
    op_value_range.to_excel(xw, sheet_name="op_value_range", index=False)
    op_step_range.to_excel(xw, sheet_name="op_step_range", index=False)
    sp_step_range.to_excel(xw, sheet_name="sp_step_range", index=False)
    combined_op_constraints.to_excel(xw, sheet_name="combined_op_constraints", index=False)

print("Saved:", OUT_XLSX)
for name, df in [("op_value_range", op_value_range),
                 ("op_step_range", op_step_range),
                 ("sp_step_range", sp_step_range),
                 ("combined_op_constraints", combined_op_constraints)]:
    print(f"  {name:26s}: {df.shape}")
```
